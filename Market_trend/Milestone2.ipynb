Importing required libraries


```python
import pandas as pd 
import matplotlib as plt
import seaborn as sns
import numpy as np

```

Loading and Inspecting DataSet

```python
df = pd.read_csv('ai_job_dataset.csv')

#shape of the dataset
print("dataset shape", df.shape)
print("")

#data info
df.info()
display(df.describe())
display(df.head())




```

Missing/null values

```python
df.isnull().sum()
```

Duplicate data

```python
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate job IDs:", df['job_id'].duplicated().sum())
```

Data by category

```python
categorical_columns = [
    'salary_currency',
    'experience_level',
    'employment_type',
    'company_location',
    'company_size',
    'employee_residence',
    'education_required',
    'industry'
]

for col in categorical_columns:
    print(f"\n{df[col].value_counts()}:")
```

Inconsistant values check (checks to make sure that there arent abnormal values such as negative years of experience or 100 years of experience)

```python
numerical_columns = [
    'salary_usd',
    'remote_ratio',
    'years_experience',
    'job_description_length',
    'benefits_score'

]

for col in numerical_columns:
    print(f"\n{col}:")
    print('minimum:', df[col].min())
    print('maximum:', df[col].max())
```

The data showed posting_date and application_deadline as a string which isnt ideal. Converting the str to datetime64 datatype. 

```python

df['posting_date'] = pd.to_datetime(df['posting_date'])
df['application_deadline'] = pd.to_datetime(df['application_deadline'])

print(df['posting_date'].head())
print(df['application_deadline'].head())
```
